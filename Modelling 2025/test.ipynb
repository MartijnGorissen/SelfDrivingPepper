```python
from ultralytics import YOLO
import torch

# 1. Load your existing model
model_path = 'obj_model.pt'
print(f"Loading pre-trained model from: {model_path}")
model = YOLO(model_path)
print("Model loaded successfully.")

# 2. Define path to your dataset configuration YAML file
data_yaml_path = 'data.yaml' 

# 3. Define training arguments

epochs_to_train = 50 
image_size = 640
batch_size = 8
run_name = 'finetune_run_with_new_objects'

print("Starting training...")
results = model.train(
    data=data_yaml_path,
    epochs=epochs_to_train,
    imgsz=image_size,
    batch=batch_size,
    name=run_name,
    workers=6
)
print("Training finished.")
print(f"Results saved to: {results.save_dir}")
```

```python
import cv2
from ultralytics import YOLO
import os
import time # Optional: for timing
import datetime

# --- Configuration ---
MODEL_PATH = 'runs/detect/finetune_run_with_new_objects3/weights/best.pt' # Path to your fine-tuned model
INPUT_VIDEO_PATHS = ['input videos/front.mp4'] # <<< CHANGE THIS to your input video file
OUTPUT_VIDEO_PATH = f'output videos/output_video_annotated_{datetime.datetime.now().strftime("%Y%m%d_%H%M%S")}.mp4' # <<< CHANGE THIS (optional) name for the output video
CONFIDENCE_THRESHOLD = 0.5

# --- Check if input files exist ---
if not os.path.exists(MODEL_PATH):
    print(f"Error: Model file not found at {MODEL_PATH}")
    exit()
# if not os.path.exists(INPUT_VIDEO_PATH):
#     print(f"Error: Input video file not found at {INPUT_VIDEO_PATH}")
#     exit()

for INPUT_VIDEO_PATH in INPUT_VIDEO_PATHS:
    print(f"Loading model from {MODEL_PATH}...")
    try:
        # Load your fine-tuned YOLO model
        model = YOLO(MODEL_PATH)
        print("Model loaded successfully!")

        # Open the input video file
        cap = cv2.VideoCapture(INPUT_VIDEO_PATH)
        if not cap.isOpened():
            print(f"Error: Could not open video file: {INPUT_VIDEO_PATH}")
            exit()

        # Get video properties (width, height, frames per second)
        frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        fps = cap.get(cv2.CAP_PROP_FPS)
        print(f"Input video properties: {frame_width}x{frame_height} @ {fps:.2f} FPS")

        # Define the codec and create VideoWriter object
        # Common codecs: 'mp4v' for .mp4, 'XVID' for .avi
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')
        out = cv2.VideoWriter(OUTPUT_VIDEO_PATH, fourcc, fps, (frame_width, frame_height))
        print(f"Output video will be saved to: {OUTPUT_VIDEO_PATH}")

        # --- Process Video Frame by Frame ---
        print("Starting video processing...")
        start_time = time.time()
        frame_count = 0

        # Use model.predict with stream=True for efficient frame-by-frame processing
        # You can add other prediction arguments like conf, iou, classes etc. here
        # e.g., model.predict(source=INPUT_VIDEO_PATH, stream=True, conf=0.5)
        for results in model.predict(source=INPUT_VIDEO_PATH, conf=CONFIDENCE_THRESHOLD, stream=True, verbose=False):
            frame_count += 1

            # results object contains detections for the current frame
            # Use the .plot() method which conveniently returns the frame with detections drawn
            annotated_frame = results.plot() # This handles drawing boxes, labels, confidences!

            # Write the annotated frame to the output video
            out.write(annotated_frame)

            # Optional: Display the annotated frame live (requires a GUI environment)
            # cv2.imshow("YOLOv8 Live Detection", annotated_frame)
            # if cv2.waitKey(1) & 0xFF == ord('q'): # Press 'q' to quit live view
            #    print("Quitting live view.")
            #    break

            if frame_count % 50 == 0: # Print progress every 50 frames
                print(f" Processed {frame_count} frames...")

        # --- Cleanup ---
        end_time = time.time()
        print(f"Finished processing {frame_count} frames.")
        print(f"Total processing time: {end_time - start_time:.2f} seconds")

        # Release video capture and writer objects
        cap.release()
        out.release()
        # Close any OpenCV display windows if opened
        # cv2.destroyAllWindows()

        print(f"Annotated video saved successfully to {OUTPUT_VIDEO_PATH}")

    except Exception as e:
        print(f"An error occurred during processing: {e}")
        import traceback
        traceback.print_exc()
        # Ensure resources are released even if an error occurs
        if 'cap' in locals() and cap.isOpened():
            cap.release()
        if 'out' in locals() and out.isOpened():
            out.release()
        # cv2.destroyAllWindows()
```
